```python
def load_and_filter_data(file_path, counties):
    """
    Loads a dataset from a CSV file and filters it for specified counties
    and 'Approved' application status.

    Args:
        file_path (str): The path to the CSV file.
        counties (list): A list of county names to filter by.

    Returns:
        pandas.DataFrame: The filtered DataFrame.
    """
    df = pd.read_csv(file_path)
    filtered_df = df[(df['CNTY_NAME'].isin(counties)) & (df['APPLICATION_STATUS'] == 'Approved')]
    print(f"Data loaded and filtered for counties: {', '.join(counties)} with 'Approved' status.")
    return filtered_df
```
